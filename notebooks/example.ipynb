```python
%load_ext autoreload
%autoreload 2

from pywerfl import loader, viz
import matplotlib.pyplot as plt
```

```python
run = loader.load_run(647)
```

```python
print(run.metadata) # Print metadata dict for run
print([item for item in dir(run) if not item.startswith("__")]) # Properties of run to access, with e.g. `run.tables`
```

```python
# Draw building with taps marked and labeled
%matplotlib inline
viz.plot_tap_locations()
plt.show()
```

```python
# Plot building mapped time series statistics: mean, std, and min pressures, as well as the BLUE estimated (expected value) min pressure (peak suction) map
%matplotlib inline
viz.plot_pressure_map(run, "mean", cp_range=(-2,1))
plt.show()
viz.plot_pressure_map(run, "std", cp_range=(0,1))
plt.show()
viz.plot_pressure_map(run, "min", show_points=True, cp_range=(-8,8))
plt.show()
viz.plot_pressure_map(run, "exp_blue_min", show_points=False, cp_range=(-8,8))
plt.show()
```

```python
# Generate and save an animation; requires FFMPEG be installed and on your system PATH.
# To run this without FFMPEG, remove the argument `writer="ffmpeg"`, and change `anim.mp4` to `anim.gif`
%matplotlib inline
anim = viz.animate_pressure_map(run, max_frames=900, show_points=True, interval_ms=33, cp_range=(-5,5))
anim.save("anim.mp4", writer="ffmpeg")
```
